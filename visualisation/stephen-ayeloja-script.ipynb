```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
# Load dataset

diabetes = pd.read_csv('diabetes.csv')
```

```python
diabetes.head()
```

```python
# plot scatter plot for blood pressure vs bmi

sns.scatterplot(data=diabetes, x='BloodPressure', y='BMI')
plt.title('Blood Pressure vs BMI')
plt.xlabel('Blood Pressure')
plt.ylabel('BMI')   
```

```python
diabetes[['BloodPressure', 'BMI']].corr()
```

```python
#stephen ayeloja's additional visualizations
sns.pairplot(diabetes, hue='Outcome')
plt.suptitle('Pairplot of Diabetes Dataset', y=1.02)
plt.show()
```
